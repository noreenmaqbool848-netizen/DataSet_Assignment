```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Step 1: Import cleaned dataset from Assignment 2
df = pd.read_csv("cleaned_iris.csv")


df.head()
```

```python
sns.histplot(df['sepal length (cm)'], kde=True)
plt.title("Sepal Length Distribution")
plt.show()

print("Insight: Sepal lengths are mostly between 5.5 and 6.0 cm.")
```

```python
sns.scatterplot(x=df['sepal length (cm)'], y=df['sepal width (cm)'])
plt.title("Sepal Length vs Sepal Width")
plt.show()

print("Insight: A slight inverse relationship between the two sepal features.")
```

```python
sns.boxplot(x=df['target'], y=df['petal length (cm)'])
plt.title("Petal Length by Species")
plt.show()

print("Insight: Species 2 has the longest petals, clearly separating groups.")
```

```python
sns.pairplot(df, hue="target")
plt.show()

print("Insight: Species form clear clusters, especially based on petal features.")
```

```python
sns.heatmap(df.corr(), annot=True, cmap="coolwarm")
plt.title("Correlation Heatmap")
plt.show()

print("Insight: Petal length and width are strongly correlated.")
```
