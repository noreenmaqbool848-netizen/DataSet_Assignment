```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

df = pd.read_csv("cleaned_iris.csv")

df.head()
```

```python
corr = df.corr()
```

```python
plt.figure(figsize=(7,5))
sns.heatmap(corr, annot=True, cmap="coolwarm")
plt.title("Correlation Heatmap")
plt.show()
```

```python
target_corr = corr['target'].sort_values(ascending=False)
target_corr
```

```python
print("Report\n"
      "The three features most strongly related to the target variable are:\n\n"
      " 1). Petal length (strongest positive correlation)\n\n"
      " 2). Petal width\n\n"
      " 3). Sepal length\n"
      "These features show clear separation between flower species, especially petal measurements.")
```
