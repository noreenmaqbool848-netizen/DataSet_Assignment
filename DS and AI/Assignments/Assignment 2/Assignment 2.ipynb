```python
import pandas as pd
from sklearn.datasets import load_iris
```

```python
iris = load_iris(as_frame=True)
df = iris.frame.copy()
```

```python
print("Before Cleaning:")
print("Shape:", df.shape)
print("Missing values:\n", df.isnull().sum())
print("Data Types:\n", df.dtypes)
```

```python
# Step 1: Remove duplicates (if any)
df.drop_duplicates(inplace=True)
```

```python
# Step 2: Convert target column to categorical type
df["target"] = df["target"].astype("category")
```

```python
print("\nAfter Cleaning:")
print("Shape:", df.shape)
print("Missing values:\n", df.isnull().sum())
print("Data Types:\n", df.dtypes)
```

```python
# Create a comparison summary
before = {
    "Rows": [150],
    "Missing Values Total": [0],
    "Duplicates": [0],
    "Target dtype": ["int64"]
}

after = {
    "Rows": [df.shape[0]],
    "Missing Values Total": [df.isnull().sum().sum()],
    "Duplicates": ["Removed if existed"],
    "Target dtype": [str(df["target"].dtype)]
}

before_after_df = pd.DataFrame({"Before": before, "After": after})
before_after_df
```

```python
# Save cleaned dataset for future assignments
df.to_csv("cleaned_iris.csv", index=False)

print("Cleaned dataset saved as cleaned_iris.csv")
```
